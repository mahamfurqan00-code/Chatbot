```python
import re
from difflib import SequenceMatcher

def normalize_text(text):
    text = text.lower().strip()
    text = re.sub(r"[^\w\s]", "", text)
    text = re.sub(r"\s+", " ", text)
    return text


def find_best_match(user_question, faq):
    user_question = normalize_text(user_question)
    best_key, best_score = None, 0

    for question in faq:
        normalized_question = normalize_text(question)
        if normalized_question in user_question or user_question in normalized_question:
            return question, 1.0

        score = SequenceMatcher(None, user_question, normalized_question).ratio()
        if score > best_score:
            best_key, best_score = question, score

    return best_key, best_score
```

```python
THRESHOLD = 0.6

def get_answer(user_input, faq):
    key, score = find_best_match(user_input, faq)
    normalized_input = normalize_text(user_input)

    # Exact normalized match
    for question, answer in faq.items():
        if normalize_text(question) == normalized_input:
            return answer

    # Best fuzzy match
    if key is not None and score >= THRESHOLD:
        return faq[key]

    # No suitable answer
    return "Sorry, I couldn't find an answer to that question."
```

```python
print("Hi! Ask me a question (type 'quit' to exit).")

while True:
    user_input = input("You: ").strip()

    if user_input.lower() == "quit":
        print("Bot: Goodbye!")
        break

    answer = get_answer(user_input, faq)
    print("Bot:", answer)
```
